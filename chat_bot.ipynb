```python
from openai import OpenAI
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

# The OpenAI SDK pointed at Groq's URL — this is how you reach the Responses API
client = OpenAI(
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
)

MODEL = "openai/gpt-oss-120b"
```

```python
# This function prompts the user for input and returns the input string.

def get_user_input():
    user_input = input("Enter your question or request: ")
    return user_input

# This function takes a list of messages and sends them to the OpenAI API to get a response from the bot. It then displays the response in Markdown format.
def get_bot_response(messages):
    response = client.chat.completions.create(
        model=MODEL,
        messages=messages
    )
    report = response.choices[0]
    display(Markdown(report.message.content))
    return report
```

```python
# Call the functions to get user input and bot response, and store the response in a variable.
response = get_bot_response(get_user_input())
response
```

```python
# Check the response object to see what it contains. You can access the message content using
response.message
```

#### Chat BOT 

```python
# This loop allows the user to have a conversation with the bot. It continues until the user types "exit" or "quit".
messages = []
while True:
    user_input = get_user_input()
    messages.append({"role": "user", "content": user_input})
    if user_input.lower() in ["exit", "quit"]:
        print("Exiting the chat. Goodbye!")
        break
    response = get_bot_response(messages)
    messages.append({"role": "assistant", "content": response.message.content})
```

### Stream Messages

```python
messages = [{
    'role':'user',
    'content':'What a 1 sentence description of a fake data base?'
}]

stream = client.chat.completions.create(
    model=MODEL,
    messages=messages,
    stream=True,
    temperature=0.7
)

for i, chunk in enumerate(stream, start=1):
    delta = chunk.choices[0].delta
    finish_reason = chunk.choices[0].finish_reason

    delta_content = getattr(delta, "content", None)
    if delta_content is None or delta_content == "":
        continue

    print(f"EVENT {i}")
    print("type:", type(delta).__name__)
    print("content:", delta_content)
    print("finish_reason:", finish_reason)
    print("-" * 60)

# for chunk in stream:
#     print(chunk.choices[0],end="", flush=True)
```
